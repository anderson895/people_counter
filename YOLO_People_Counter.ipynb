## 2️⃣ Import Libraries

```python
import cv2
from ultralytics import YOLO
```

## 3️⃣ Load YOLO Model

```python
model = YOLO('yolov5s.pt')  # Small YOLOv5 pre-trained on COCO dataset
```

## 4️⃣ Open the Camera

```python
cap = cv2.VideoCapture(0)
if not cap.isOpened():
    print('❌ Cannot open camera')
    exit()
```

## 5️⃣ Main Loop: Detect People in Real-Time

```python
while True:
    ret, frame = cap.read()
    if not ret:
        break
    
    results = model(frame, verbose=False)[0]
    person_count = 0
    
    if results.boxes is not None:
        for box in results.boxes:
            cls_id = int(box.cls[0])
            conf = float(box.conf[0])
            class_name = model.names[cls_id]
            if class_name == 'person':
                person_count += 1
                x1, y1, x2, y2 = map(int, box.xyxy[0])
                cv2.rectangle(frame, (x1, y1), (x2, y2), (0, 255, 0), 2)
                cv2.putText(frame, f'Person {conf:.2f}', (x1, y1 - 10), cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0, 255, 0), 2)
    cv2.putText(frame, f'Total Persons: {person_count}', (20, 40), cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 0, 255), 3)
    cv2.imshow('YOLO People Counter', frame)
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```

### ✅ Notes
- Press **q** to quit the live camera window.
- The program highlights detected people with green bounding boxes and shows confidence scores.
- The total number of detected people is shown in red at the top-left corner.
